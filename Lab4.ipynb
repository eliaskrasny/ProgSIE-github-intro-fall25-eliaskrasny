```python
#import
import seaborn as sns
iris = sns.load_dataset('iris')
```

```python
#20
#What is minimum value of sepal_width for "virginica" variety in the iris dataset?
virginica_df = iris[iris['species'] == 'virginica']
min_sepal_width_virginica = virginica_df['sepal_width'].min()
print(f"The minimum sepal width for 'virginica' variety is: {min_sepal_width_virginica}")
```

```python
#21
#Construct a dataframe where petal_length is greater than the mean petal_length of whole iris dataset.
#How many observations are there in this dataframe?

mean_petal_length = iris['petal_length'].mean()
petal_length_greater_than_mean_df = iris[iris['petal_length'] > mean_petal_length]

print(f"Number of observations where petal_length is greater than the mean: {len(petal_length_greater_than_mean_df)}")
```

```python
#22
# Check if there are 43 versicolor
versicolor_count = iris[iris['species'] == 'versicolor'].shape[0]
print(f"Number of versicolor observations: {versicolor_count}")

# Check the mean of sepal_length
mean_sepal_length = iris['sepal_length'].mean()
print(f"Mean sepal_length: {mean_sepal_length}")

# Check the number of unique species
num_species = iris['species'].nunique()
print(f"Number of unique species: {num_species}")

# Check if there is 0 virginica
virginica_count = iris[iris['species'] == 'virginica'].shape[0]
print(f"Number of virginica observations: {virginica_count}")
```

```python
#23
# Calculate the Pearson correlation for each pair
correlation_sepal_length_width = iris['sepal_length'].corr(iris['sepal_width'], method='pearson')
correlation_petal_length_sepal_length = iris['petal_length'].corr(iris['sepal_length'], method='pearson')
correlation_sepal_length_petal_width = iris['sepal_length'].corr(iris['petal_width'], method='pearson')
correlation_petal_length_petal_width = iris['petal_length'].corr(iris['petal_width'], method='pearson')

# Print the correlations
print(f"Pearson correlation between sepal_length and sepal_width: {correlation_sepal_length_width}")
print(f"Pearson correlation between petal_length and sepal_length: {correlation_petal_length_sepal_length}")
print(f"Pearson correlation between sepal_length and petal_width: {correlation_sepal_length_petal_width}")
print(f"Pearson correlation between petal_length and petal_width: {correlation_petal_length_petal_width}")

# Find the maximum correlation among the pairs
correlations = {
    'sepal_length, sepal_width': abs(correlation_sepal_length_width),
    'petal_length, sepal_length': abs(correlation_petal_length_sepal_length),
    'sepal_length, petal_width': abs(correlation_sepal_length_petal_width),
    'petal_length, petal_width': abs(correlation_petal_length_petal_width)
}

highest_correlation_pair = max(correlations, key=correlations.get)
print(f"\nThe variable pair with the highest Pearson correlation is: {highest_correlation_pair}")
```

```python
#24
# Filter for setosa and virginica species
setosa_df = iris[iris['species'] == 'setosa']
virginica_df = iris[iris['species'] == 'virginica']

# Check the mean value of petal_length of setosa
mean_petal_length_setosa = setosa_df['petal_length'].mean()
print(f"Mean petal_length of setosa: {mean_petal_length_setosa:.2f}")

# Check if setosa has the largest average sepal_width
mean_sepal_width_setosa = setosa_df['sepal_width'].mean()
mean_sepal_width_versicolor = iris[iris['species'] == 'versicolor']['sepal_width'].mean()
mean_sepal_width_virginica = virginica_df['sepal_width'].mean()
print(f"Average sepal_width of setosa: {mean_sepal_width_setosa}")
print(f"Average sepal_width of versicolor: {mean_sepal_width_versicolor}")
print(f"Average sepal_width of virginica: {mean_sepal_width_virginica}")
print(f"Setosa has the largest average sepal_width: {mean_sepal_width_setosa > mean_sepal_width_versicolor and mean_sepal_width_setosa > mean_sepal_width_virginica}")


# Check the standard deviation of sepal_length of setosa
std_sepal_length_setosa = setosa_df['sepal_length'].std()
print(f"Standard deviation of sepal_length of setosa: {std_sepal_length_setosa:.2f}")

# Check if virginica has largest sepal_length
max_sepal_length_setosa = setosa_df['sepal_length'].max()
max_sepal_length_versicolor = iris[iris['species'] == 'versicolor']['sepal_length'].max()
max_sepal_length_virginica = virginica_df['sepal_length'].max()
print(f"Largest sepal_length in setosa: {max_sepal_length_setosa}")
print(f"Largest sepal_length in versicolor: {max_sepal_length_versicolor}")
print(f"Largest sepal_length in virginica: {max_sepal_length_virginica}")
print(f"Virginica has the largest sepal_length: {max_sepal_length_virginica > max_sepal_length_setosa and max_sepal_length_virginica > max_sepal_length_versicolor}")
```

```python
import seaborn as sns
titanic = sns.load_dataset('titanic')
```

```python
#25
# Filter for 2nd class passengers who survived
survived_2nd_class_df = titanic[(titanic['pclass'] == 2) & (titanic['survived'] == 1)]

# Count the number of observations
num_survived_2nd_class = len(survived_2nd_class_df)

print(f"The number of passengers in the 2nd class who survived is: {num_survived_2nd_class}")
```

```python
#26
# Filter for 1st class passengers who did not survive
did_not_survive_1st_class_df = titanic[(titanic['pclass'] == 1) & (titanic['survived'] == 0)]

# Find the minimum and maximum age
min_age_did_not_survive_1st_class = did_not_survive_1st_class_df['age'].min()
max_age_did_not_survive_1st_class = did_not_survive_1st_class_df['age'].max()

print(f"The age range for 1st class passengers who did not survive is: {min_age_did_not_survive_1st_class} to {max_age_did_not_survive_1st_class}")
```

```python
# Filter for female passengers
female_passengers = titanic[titanic['sex'] == 'female']

# Count female passengers who survived
survived_females = female_passengers[female_passengers['survived'] == 1].shape[0]

# Count total female passengers
total_females = female_passengers.shape[0]

# Calculate the percentage of female survivors
percentage_female_survivors = (survived_females / total_females) * 100

print(f"The percentage of female passengers who survived is: {percentage_female_survivors:.0f}%")
```

```python
# Check for missing values in the specified columns
missing_sex = titanic['sex'].isnull().sum()
missing_embarked = titanic['embarked'].isnull().sum()
missing_deck = titanic['deck'].isnull().sum()
missing_age = titanic['age'].isnull().sum()

print(f"Missing values in 'sex': {missing_sex}")
print(f"Missing values in 'embarked': {missing_embarked}")
print(f"Missing values in 'deck': {missing_deck}")
print(f"Missing values in 'age': {missing_age}")
```

```python
# Calculate survival rate by passenger class
survival_rate_by_class = titanic.groupby('pclass')['survived'].mean().sort_values(ascending=False)
print(f"Survival rate by passenger class:\n{survival_rate_by_class}\n")

# Calculate survival rate by embarkation point
survival_rate_by_embarked = titanic.groupby('embarked')['survived'].mean().sort_values(ascending=False)
print(f"Survival rate by embarkation point:\n{survival_rate_by_embarked}\n")

# Calculate correlation between survival and passenger class
correlation_survival_pclass = titanic['survived'].corr(titanic['pclass'])
print(f"Correlation between survival and passenger class: {correlation_survival_pclass}\n")

# Calculate correlation between survival and fare
correlation_survival_fare = titanic['survived'].corr(titanic['fare'])
print(f"Correlation between survival and fare: {correlation_survival_fare}\n")

# Evaluate the statements
print("Evaluating the statements:")
# Statement 1: Survival rate is highest in 1st class passengers
is_highest_in_1st_class = survival_rate_by_class.index[0] == 1
print(f"Survival rate is highest in 1st class passengers: {is_highest_in_1st_class}")

# Statement 2: Survival rate is highest for passengers who embarked at S = Southampton
is_highest_embarked_S = survival_rate_by_embarked.index[0] == 'S'
print(f"Survival rate is highest for passengers who embarked at S = Southampton: {is_highest_embarked_S}")

# Statement 3: Survival rate is highest for passengers who embarked at C = Cherbourg
is_highest_embarked_C = survival_rate_by_embarked.index[0] == 'C'
print(f"Survival rate is highest for passengers who emabarked at C = Cherbourg: {is_highest_embarked_C}")

# Statement 4: Survival has positive correlation with passenger class
has_positive_correlation_pclass = correlation_survival_pclass > 0
print(f"Survival has positive correlation with passenger class: {has_positive_correlation_pclass}")

# Statement 5: Survival has negative correlation with fare
has_negative_correlation_fare = correlation_survival_fare < 0
print(f"Survival has negative correlation with fare: {has_negative_correlation_fare}")
```
