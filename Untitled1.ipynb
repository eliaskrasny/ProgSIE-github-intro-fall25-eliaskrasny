```python
import pandas as pd

df = pd.read_csv('/content/spam.csv')
display(df.head())
```

```python
#27
spamonly = df[df['v1'] == 'spam']

duplicates_count = spamonly.duplicated().sum()
print(f"Number of duplicate entries in spamonly: {duplicates_count}")

display(spamonly.head())
```

```python
#28
spamonly_no_duplicates = spamonly.drop_duplicates()
total_unique_spam = len(spamonly_no_duplicates)
print(f"Total number of unique spam SMS: {total_unique_spam}")
```

```python
#29
df_no_duplicates = df.drop_duplicates()
df_no_duplicates['message_length'] = df_no_duplicates['v2'].apply(len)
average_length = round(df_no_duplicates['message_length'].mean(), 2)
print(f"Average message length in the duplicate-free full SMS dataset: {average_length} characters")
```

```python
import re

def count_sentences(text):
    # Split by one or more sentence-ending punctuation marks
    # Filter out any empty strings that result from the split (e.g., trailing punctuation)
    sentences = [s for s in re.split(r'[.!?]+', text.strip()) if s]
    return len(sentences)

df_no_duplicates_copy = df_no_duplicates.copy()
df_no_duplicates_copy['sentences_count'] = df_no_duplicates_copy['v2'].apply(count_sentences)
max_sentences_count = df_no_duplicates_copy['sentences_count'].max()

print(f"Maximum sentences count in the duplicate-free full SMS dataset is: {max_sentences_count}")
```

```python
import re

def count_words(text):
    words = re.findall(r'\b\w+\b', text.lower())
    return len(words)

df_no_duplicates_copy['words_count'] = df_no_duplicates_copy['v2'].apply(count_words)
min_words_count = df_no_duplicates_copy['words_count'].min()

print(f"Minimum words count in the duplicate-free full SMS dataset: {min_words_count}")
```

```python
import re

def count_capital_letters(text):
    return sum(1 for char in text if char.isupper())

# Get the first five messages from the duplicate-free DataFrame
first_five_messages = df_no_duplicates['v2'].head(5)

# Apply the function to count capital letters for each of these messages
capital_counts = first_five_messages.apply(count_capital_letters)

# Find the highest number of capital letters
highest_capital_letters = capital_counts.max()

print(f"The highest number of capital letters in the first five duplicate-free SMS is: {highest_capital_letters}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import re

def count_capital_letters(text):
    return sum(1 for char in text if char.isupper())

# Add capital_letters_count to df_no_duplicates_copy
df_no_duplicates_copy['capital_letters_count'] = df_no_duplicates_copy['v2'].apply(count_capital_letters)

# Filter for spam messages from the duplicate-free dataset with all features
spam_df_features = df_no_duplicates_copy[df_no_duplicates_copy['v1'] == 'spam']

# Define the features for which to create histograms
features_to_plot = ['sentences_count', 'words_count', 'message_length', 'capital_letters_count']

# Create histograms
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 10))
axes = axes.flatten() # Flatten the 2x2 array of axes for easy iteration

for i, feature in enumerate(features_to_plot):
    sns.histplot(spam_df_features[feature], kde=True, ax=axes[i])
    axes[i].set_title(f'Histogram of {feature} for Spam SMS')
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
